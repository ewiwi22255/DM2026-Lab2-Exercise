## 1. Model Development (10 pts Required)

### 1.1 Preprocessing Steps

First, the tables are merged using `post_id`. The data is then split into training and test sets using the `set` column. Text is processed via a `clean_text` function, which converts text to lowercase, removes URLs and @mentions, and collapses excessive whitespace. The distribution of the six labels is verified as follows: joy (875), anger (674), sadness (628), disgust (454), surprise (441), and fear (288). URLs and user mentions are removed because they do not contribute to emotion detection.

### 1.2 Feature Engineering Steps

A validation set is created by reserving 20% of the training data, using stratification to ensure the proportions of the six emotions remain consistent across both sets. Text is converted into features using TF-IDF, with parameters set to `min_df=2` (retaining terms that appear in at least two posts) and `ngram_range=(1, 2)` (capturing both unigrams and bigrams). The TF-IDF vectorizer is fitted *only* on the training data, while the validation and test sets are transformed using the fitted vectorizer; this prevents data leakage (i.e., the model "peeking" at the test/validation data). Based on Section 6 and Exercise 9 of the Master notebook, TF-IDF was identified as the most effective method. A 20% validation set is used because the model has not encountered this data during training; furthermore, since the Kaggle test set lacks labels, the validation set provides a reliable metric for performance.

### 1.3 Explanation of Your Model

Two models were trained: Linear SVM (with `C=1.0`) and Multinomial Naive Bayes. The Linear SVM achieved a higher validation macro F1 (0.4898) compared to the Multinomial Naive Bayes (0.2977), leading to the selection of the Linear SVM. The F1-scores for the Linear SVM across the different emotions are: joy 0.60, sadness 0.54, fear 0.53, surprise 0.51, anger 0.45, and disgust 0.31. For the "fear" category, the precision is 0.71 and the recall is 0.42. Some texts can be misclassified as a different emotion. # **Competition Code**

## 1. Preprocessing Steps

```python
### Add the code related to the preprocessing steps in cells inside this section
import re
import pandas as pd

DATA_DIR = "data/kaggle"

#  1. Load the three files 
posts = pd.read_json(f"{DATA_DIR}/final_posts.json")         # post_id, text
ident = pd.read_csv(f"{DATA_DIR}/data_identification.csv")   # post_id, set
labels = pd.read_csv(f"{DATA_DIR}/emotion.csv")              # post_id, label (train only)

#  2. Merge them on post_id 
df = posts.merge(ident, on="post_id")
df = df.merge(labels, on="post_id", how="left")

#  3. Split into train and test 
train_df = df[df["set"] == "train"].copy()
test_df = df[df["set"] == "test"].copy()
#  4. Clean the text 
def clean_text(s: str) -> str:
    s = s.lower()
    s = re.sub(r"http\S+|www\.\S+", " ", s)
    s = re.sub(r"@\w+", " ", s)
    return re.sub(r"\s+", " ", s).strip()

train_df["clean"] = train_df["text"].map(clean_text)
test_df["clean"] = test_df["text"].map(clean_text)

#  5. Check the result 
print(train_df.shape, test_df.shape)
print(train_df["label"].value_counts())
```

## 2. Feature Engineering Steps

```python
### Add the code related to the feature engineering steps in cells inside this section
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

SEED = 42

#  1. Hold out 20% of the labelled posts as a validation set 
X_tr_text, X_val_text, y_tr, y_val = train_test_split(
    train_df["clean"], train_df["label"],
    test_size=0.2, stratify=train_df["label"], random_state=SEED)

#  2. TF-IDF features: fit on the training part only 
tfidf = TfidfVectorizer(min_df=2, ngram_range=(1, 2))
X_tr = tfidf.fit_transform(X_tr_text)
X_val = tfidf.transform(X_val_text)
X_test = tfidf.transform(test_df["clean"])

print("train:", X_tr.shape, "| validation:", X_val.shape, "| test:", X_test.shape)
print("vocabulary size:", len(tfidf.vocabulary_))
```

## 3. Model Implementation Steps

```python
### Add the code related to the model implementation steps in cells inside this section
from sklearn.svm import LinearSVC
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import f1_score, classification_report

#  1. Linear SVM: train on the training part, score on validation 
svm = LinearSVC(random_state=SEED)
svm.fit(X_tr, y_tr)
y_pred_svm = svm.predict(X_val)
print("Linear SVM  validation macro F1:", round(f1_score(y_val, y_pred_svm, average="macro"), 4))
print(classification_report(y_val, y_pred_svm))

#  2. Baseline for comparison: Multinomial Naive Bayes 
nb = MultinomialNB()
nb.fit(X_tr, y_tr)
y_pred_nb = nb.predict(X_val)
print("Naive Bayes validation macro F1:", round(f1_score(y_val, y_pred_nb, average="macro"), 4))

#  3. Refit on all labelled posts, predict the test set, save the predictions 
X_all = tfidf.fit_transform(train_df["clean"])
X_test_all = tfidf.transform(test_df["clean"])
final_model = LinearSVC(random_state=SEED).fit(X_all, train_df["label"])

submission = pd.DataFrame({"post_id": test_df["post_id"],
                           "label": final_model.predict(X_test_all)})
submission.to_csv("data/kaggle/my_submission.csv", index=False)
print(submission.shape)
print(submission["label"].value_counts())
```
